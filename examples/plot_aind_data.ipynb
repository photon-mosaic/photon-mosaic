# Plotting sample data from AIND

In this simple example, we showcase `photon-mosaic`'s plotting functionality to explore some publicly available data from the Allen Institute for Neural Dynamics (AIND). First, we use a function from our `sample_data` module to download the data (this will take a while the first time you do it!)

```python
%matplotlib widget
from photon_mosaic.sample_data import download_aind_multiregion_data
aind_data = download_aind_multiregion_data(fields_of_view=[0,1], regions=[0, 1]) # only get first two regions
```

This real-life data is spatially complex (spanning 4 regions, each with 2 concurrently imaged fields of view (FOVs) at differing depths). For the purposes of this example, we restrict ourselves to plotting it. First we show side by side plot of both FOVs in the first region:

```python
from photon_mosaic.widgets.series import plot_imaging_series

region_0_data = {
    f"field of view {fov}": imaging
    for (fov, region), imaging in aind_data.items()
    if region == 0
}

plot_imaging_series(
    region_0_data,
    backend="ipywidgets",
    colormap="plasma",
)
```

Same plot, different region of interest:

```python

region_1_data = {
    f"channel {channel}": imaging
    for (channel, region), imaging in aind_data.items()
    if region == 1
}

plot_imaging_series(
    region_1_data,
    backend="ipywidgets",
    colormap="plasma",
)
```

First FOV from both regions alongside each other:

```python
fovs_across_regions = {
    f"FOV in {region}": imaging
    for (fov, region), imaging in aind_data.items()
    if fov == 0
}

plot_imaging_series(
    fovs_across_regions,
    backend="ipywidgets",
    colormap="plasma",
)
```

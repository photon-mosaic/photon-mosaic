# Plotting sample data from AIND

In this simple example, we showcase `photon-mosaic`'s plotting functionality to explore some publicly available data from the Allen Institute for Neural Dynamics (AIND). First, we use a function from our `sample_data` module to download the data (this will take a while the first time you do it!)

```python
%matplotlib widget
from photon_mosaic.sample_data import download_aind_multiregion_data
# get central plane in both channels from first two regions
aind_data = download_aind_multiregion_data(channels=[1, 2], regions=[0, 1])
```

While this real-life data is spatially complex (spanning multiple regions and channels), it is also quite short in time (20 frames), which may break the assumptions of some algorithms. For the purposes of this example, we therefore restrict ourselves to plotting it. First we show side by side plot of both channels in the first region:

```python
from photon_mosaic.widgets.series import plot_imaging_series

region_0_data = {
    f"channel {channel}": imaging
    for (channel, region), imaging in aind_data.items()
    if region == 0
}

plot_imaging_series(
    region_0_data,
    backend="ipywidgets",
    colormap="plasma",
)
```

Same plot, different region:

```python

region_1_data = {
    f"channel {channel}": imaging
    for (channel, region), imaging in aind_data.items()
    if region == 1
}

plot_imaging_series(
    region_1_data,
    backend="ipywidgets",
    colormap="plasma",
)
```

Signal channels from both regions alongside each other:

```python
signal_data_across_regions = {
    f"channel 2 in {region}": imaging
    for (channel, region), imaging in aind_data.items()
    if channel == 2
}

plot_imaging_series(
    {key: value for key, value in aind_data.items() if key[0] == 2},
    backend="ipywidgets",
    colormap="plasma",
)
```

# Plotting sample data from AIND

In this simple example, we showcase `photon-mosaic`'s plotting functionality to explore some publicly available data from the Allen Institute for Neural Dynamics (AIND). First, we use a function from our `sample_data` module to download the data

```python
%matplotlib widget
from photon_mosaic.sample_data import download_aind_multiregion_data

# get central plane in both channels from first two regions
channel_1_region_0, channel_1_region_1, channel_2_region_0, channel_2_region_1 = download_aind_multiregion_data(channels=[1,2],regions=[0,1])
```

While this real-life data is spatially complex (spanning multiple regions and channels), it is also quite short in time (20 frames), which may break the assumptions of some algorithms. For the purposes of this example, we therefore restrict ourselves to plotting it. First we show side by side plot of both channels in the first region:

```python
from photon_mosaic.widgets.series import plot_imaging_series
plot_imaging_series({"channel 1": channel_1_region_0, "channel 2": channel_2_region_0}, backend="ipywidgets", colormap="plasma")
```

Same plot, different region:

```python
plot_imaging_series({"channel 1": channel_1_region_1, "channel 2": channel_2_region_1}, backend="ipywidgets", colormap="plasma")
```

Signal channels from both regions alongside each other:

```python
plot_imaging_series({"region 0": channel_2_region_0, "region 1": channel_2_region_1}, backend="ipywidgets", colormap="plasma")
```

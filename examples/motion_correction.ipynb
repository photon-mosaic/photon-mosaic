```python
import photon_mosaic.extractors as pme
from photon_mosaic.preprocessing.suite2p import compute_motion_suite2p, RegisterSuite2PImaging
from photon_mosaic.preprocessing.suite2p_segmentation import detect_rois_suite2p
from pathlib import Path
```

```python
image_fp = [str(i) for i in Path("C:/Users/ariellel/data").glob("spont_*.tif")]
```

### Read data from imaging path using read_tiff_imaging

```python
image = pme.read_tiff_imaging(file_path=image_fp[0],sampling_frequency=30.0)
image
```

### Compute displacements using suit2p registration

```python
shift_image = compute_motion_suite2p(image)
```

### Register frames to dispalcement

```python
registered_image =RegisterSuite2PImaging(image, shift_image)
```

```python
import matplotlib.pyplot as plt

avg = registered_image.get_average_image()  # shape: (H, W, n_planes)

plt.imshow(avg[:, :, 0], cmap="gray")
plt.axis("off")
plt.title("Mean image")
plt.show()
```

```python

%matplotlib widget
from photon_mosaic.widgets import plot_imaging_series

plot_imaging_series(registered_image, backend="ipywidgets")
```

### Run segmentation using default segmentation parameters

```python
segmentation = detect_rois_suite2p(registered_image,shift_image.yranges[0][0], shift_image.xranges[0][0], shift_image.corrected_badframes[0][0])
```

```python
from photon_mosaic.widgets import plot_rois

plot_rois(segmentation, imaging=registered_image, backend="ipywidgets")
```
